```python
# Name: Girish Chapekar
# USN : CS23013
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import kagglehub
from kagglehub import KaggleDatasetAdapter
```

```python
# Name: Girish Chapekar
# USN : CS23013
#Step 1: Load Real Dataset
# Step 1: Load Insurance Dataset
file_path = "insurance.csv"
# Load the latest version
df = kagglehub.load_dataset(KaggleDatasetAdapter.PANDAS,"teertha/ushealthinsurancedataset",file_path)


print("First 5 records:\n", df.head())
```

```python
df.describe()
```

```python
df.info()
```

```python
df.shape
```

```python
# features and result split
x=df.drop('charges',axis=1)
y=df['charges']
```

```python
x.head()
```

```python
y.head()
```

```python
x = pd.get_dummies(x, drop_first=True)
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3, random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
#step 4: Train Linear Regression Model
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
#step 5: Prediction
y_pred = model.predict(x_test)
```

```python
#step 6:Model Evaluation
import numpy as np

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)
```

```python
print(f"\n Model Evaluation Matrices:")
print(f"Mean Absolute Error (mae):{mae}")
print(f"Mean Squared Error (mse):{mse}")
print(f"Root Mean Squared Error (rmse):{rmse}")
print(f"R squared Score(R2):{r2}")
```

```python
plt.scatter(x_test['age'], y_test, label='Actual Values')
plt.scatter(x_test['age'], y_pred, label='Predicted Values')
plt.xlabel('Age')
plt.ylabel('Charges')
plt.title('Linear Regression on Insurance Dataset')
plt.legend()
plt.show()
```
